```python
#Floristeria

import random

# -----------------------------
# 1. FUNCIÓN: CREAR INVENTARIO
# -----------------------------
def inventario():
    """Genera un inventario aleatorio de 3 flores distintas."""
    flores = ["tulipán", "lirio", "rosa", "dahlia", "margarita", "girasol", "clavel", "geranio"]
    seleccion = random.sample(flores, 3)

    inventario = {
        flor: {
            "precio": random.randint(2000, 6000),
            "cantidad": random.randint(10, 50),
            "vendidas": 0
        } for flor in seleccion
    }
    return inventario


# -----------------------------
# 2. FUNCIÓN: MOSTRAR MENÚ
# -----------------------------
def menu(inventario):
    """Muestra las flores disponibles en la floristería."""
    print("\n--- MENÚ DE FLORES DISPONIBLES - FLORISTERÍA CARLITOS ---")
    for flor, datos in inventario.items():
        print(f"{flor.capitalize():<10} | Precio: ${datos['precio']} | Disponibles: {datos['cantidad']}")
    print("----------------------------------------------------------")


# -----------------------------
# 3. FUNCIÓN: REALIZAR VENTA
# -----------------------------
def ventas(inventario, flor, cantidad, total_ganado):
    """Realiza la venta de una flor si hay stock suficiente."""
    if flor not in inventario:
        print(f"❌ {flor} no se encuentra a la venta.")
        return total_ganado

    datos = inventario[flor]
    if cantidad > datos["cantidad"]:
        print(f"⚠️ Solo hay {datos['cantidad']} unidades disponibles de {flor}.")
        return total_ganado

    # Procesar la venta
    datos["cantidad"] -= cantidad
    datos["vendidas"] += cantidad
    venta = datos["precio"] * cantidad
    total_ganado += venta

    print(f"✅ Vendidas {cantidad} unidades de {flor} por un total de ${venta}.")

    # Si se agota el stock, eliminar la flor
    if datos["cantidad"] == 0:
        print(f"🌸 {flor} se ha agotado y fue retirada del inventario.")
        del inventario[flor]

    return total_ganado


# -----------------------------
# 4. FUNCIÓN: CERRAR TIENDA
# -----------------------------
def cerrar_tienda(inventario_inicial, inventario_final, total_ganado):
    """Muestra las flores más y menos vendidas, y el total ganado."""
    print("\n--- TIENDA CERRADA ---")
    flores_vendidas = {}

    # Comparar inventarios inicial y final
    for flor, datos_iniciales in inventario_inicial.items():
        cantidad_final = inventario_final.get(flor, {"cantidad": 0})["cantidad"]
        vendidas = datos_iniciales["cantidad"] - cantidad_final
        flores_vendidas[flor] = vendidas

    # Mostrar resultados
    print("\n📊 Resumen de ventas:")
    for flor, cantidad in flores_vendidas.items():
        print(f"{flor.capitalize():<10}: {cantidad} unidad(es) vendidas")

    # Determinar más y menos vendidas
    mas_vendida = max(flores_vendidas, key=flores_vendidas.get)
    menos_vendida = min(flores_vendidas, key=flores_vendidas.get)

    print(f"\n💐 Flor más vendida: {mas_vendida} ({flores_vendidas[mas_vendida]} unidad(es))")
    print(f"🥀 Flor menos vendida: {menos_vendida} ({flores_vendidas[menos_vendida]} unidad(es))")
    print(f"\n💰 Total ganado: ${total_ganado}")
    print("Gracias por visitar la Floristería 🌷")


# -----------------------------
# 5. FUNCIÓN PRINCIPAL
# -----------------------------
def floristeria():
    """Ejecuta el flujo completo del programa."""
    inventario_inicial = inventario()
    inventario_actual = {k: v.copy() for k, v in inventario_inicial.items()}
    total_ganado = 0

    while True:
        menu(inventario_actual)
        opcion = input("\n¿Deseas comprar flores? (si/no): ").lower()

        if opcion == "no":
            break

        flor = input("Ingrese el nombre de la flor que desea vender: ").lower()
        try:
            cantidad = int(input("¿Cuántas unidades desea vender?: "))
        except ValueError:
            print("Por favor, ingresa un número válido.")
            continue

        total_ganado = ventas(inventario_actual, flor, cantidad, total_ganado)

    cerrar_tienda(inventario_inicial, inventario_actual, total_ganado)


# -----------------------------
# 6. EJECUCIÓN
# -----------------------------
if __name__ == "__main__":
    floristeria()
```
